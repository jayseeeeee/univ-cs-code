## Upload Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load .csv file
df = pd.read_csv("who_suicide_statistics.csv")

# Remove rows with missing data
df = df.dropna(subset=['suicides_no'])

df.head()
```

## Histogram

Create a histogram using chosen column 1

```python
counts, bins, _ = plt.hist(df["suicides_no"], edgecolor="black", color="red")

plt.title("Number of Total Global Suicides From 1979–2016")
plt.xlabel("Number of Suicides")
plt.ylabel("Frequency")
```

## Frequency Polygon

Create a frequence polygon using chosen column 1

```python
midpoints = (bins[:-1] + bins[1:]) / 2

plt.plot(midpoints, counts, marker="x", color="red")
plt.title("Frequency Polygon")
plt.xlabel("Number of Suicides")
plt.ylabel("Frequency")
```

## Scatter Plot

Instructions:
Use:
X = Column 1
Y = Column 2

```python
plt.scatter(df["suicides_no"], df["year"], color='red')
plt.title("Number of Global Suicides by Year")
plt.xlabel("Number of Suicides")
plt.ylabel("Year")
```

What is the range and the variance? What does it tell you about your data?
_____________

    A large variance like this indicates that suicide counts fluctuate significantly from year to year rather than staying close to a central average where the data is highly dispersed.

    A wide range like the one computed below means there is a big gap between the lowest and highest suicide counts recorded. This reinforces the dataset's characteristic of having relatively low and very high values.

```python
# Calculate variance for a column
variance = df['suicides_no'].var().round(2)
print("Variance:", variance)

# Calculate range for a column
data_range = df['suicides_no'].max() - df['suicides_no'].min()
print("Range:", data_range)
```

## Interpretation
-What does your histogram show?

    The histogram shows that the majority of reported suicide reported by each country every year hovers around less than five-thousand (5000) with suicides numbers above being exceedingly rare in contrast to the majority reported cases.

-What kind of skewness is present?

    The histogram depicts a _positive skew_ or it is _right-skewed_ since the histogram or the frequency polygons shows extremely rare occurrence of suicide cases above five-thousand (5000) which stretches the tail to the right.

-What does your scatterplot reveal?

    The scatterplot depicts a similar scenario to histogram where the majority of reported suicide cases hovers under five thousand (5000) with the added category of year, although it can be analyzed here that per increasing year, more suicide cases is reported and it can be better visualized here on which year the outliers increase or decrease.

-Is your data consistent or spread out?

    For the vast majority of the reported cases, it is vastly consistent where each reported cases fall under five thousand (5000) with the given graphical summaries. Although there are a few outliers and skewed data, they are still vastly limited compared to the majority falling under the mean.
