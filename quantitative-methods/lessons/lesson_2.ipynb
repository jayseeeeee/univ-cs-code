# **Python Dataset Manipulation**
The _sample_data.csv_ is manipulated using Python through the _pandas_ library. With this library, the csv can be cleaned, sorted, and manipulated with other various functions.

The script shows the dataset **before** manipulation.

```
import pandas as pd

df = pd.read_csv("sample_data.csv")
```

```
# Get dataset table shape
print(df.shape)
```

```
# Display first 5 row of dataset
df.head()
```

The script below shows the dataset **after** manipulation.

```
# Remove score > 100 (impossible)
df = df[df['Score'] <= 100]

# Remove rows with missing score
df = df.dropna(subset=['Score'])

# Remove duplicate rows
df = df.drop_duplicates()
print(f'Clean rows: {len(df)}')
```

```
# Show clean dataset
df.head()
```

### Organization of Data
The script below shows how the dataset can be reorganized through various functions.


```
# Sort by score (high to low)
df_organized = df.sort_values('Score', ascending=False)

# Filter by score (Section A only)
df_organized = df_organized[df_organized['Section'] == 'A']

# Rename a column
df_organized = df_organized.rename(columns={'Score' : 'QuizScore'})

# Summary of statistics
df_organized.head()
```

```
# Get the mean value of specific column
print("Mean of Quiz Score:", df_organized['QuizScore'].mean())
```

```
# Get the frequency of values in specified column
print(df_organized['Section'].value_counts())
```

```
# Show data types of each column
print(df_organized.dtypes)
```

```
# Round off data description
df_organized.describe().round(2)
```

```
# For better clarity and specific column target
df_organized['QuizScore'].describe()
```
