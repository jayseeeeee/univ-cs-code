# **Python Graphical Data Summarization**
Data can be summarized through _matplotlib_ through graphs and visualization.

```python
import matplotlib.pyplot as pt

study_hours = [1,2,3,4,4,5,6,7,7,8]
scores = [50,50,50,55,60,65,70,75,80,85]
```

### Histogram
Visual interpretation of numerical data by showing the number of data points that fall within a specified range of values.

```python
# Store histogram data
counts, bins, _ = pt.hist(study_hours, bins=4, edgecolor="black")

pt.title("Study Hours vs Score")
pt.xlabel("Study Hours")
pt.ylabel("Score")
```

### Plot
Example of Frequency Polygon

```python
midpoints = (bins[:-1] + bins[1:]) / 2

pt.plot(midpoints, counts, marker="o")
pt.title("Frequency Polygon")
pt.xlabel("Score Range")
pt.ylabel("Frequency")
```

### Scatter
Display the relationship between two continuous variables using Cartesian coordinates.

```python
pt.scatter(study_hours, scores, marker='o', color='red')
pt.title("Study Hours vs Score")
pt.xlabel("Study Hours")
pt.ylabel("Score")
```

### Pie
Data summarization through pie representation.

```python
pt.pie(scores)
pt.title("Score Graphed in Pie")
pt.xlabel("Study Hours")
pt.ylabel("Score")
```
